# Discussion

La [démonstration](./demonstration_FR.ipynb) précédente est celle faite dans la foulée de la conjecture initiale. 

La propriété du Damier lorsque $N$ est premier, découle de la propriété de congruence des coefficients binomiaux sur les lignes dont les indices sont des nombres premiers, et d'une propriété de symétrie du triangle de Pascal.

La recherche bibliographique ne nous a pas permis de trouver de travaux antérieurs:
- introduisant et étudiant de manière générale, la décomposition des coefficients binomiaux $A_{i,j}=\binom{i}{j}$ et $B^N_{i,j}=\binom{N-1-j}{N-1-i}$ à partir des termes $S^N_{i,j}$ et $W^N_{i,j}$.
- le lien entre cette décomposition, les nombres de Lucas, et les nombres pseudo-premiers de Fibonacci.

Plusieurs propriétés du damier peuvent néanmoins être mise en perspective avec certains travaux antérieurs, dont ceux de Rowland (2022), Prunescu (2022), et Cinkr and Ozturkalan (2023), 

Nous développons ici, quelques éléments de discussion en lien avec notre étude, et ces études.


## Concernant la décomposition des coefficients binomiaux

Le terme $A_{i,j}=A_{j,i}=\binom{i}{j}$ est le seul qui est indépendant de $N$. 

La propriété, pour $0\leq j \leq i \leq N-1$, $A_{i,j}={1\over2}\left(S^N_{i,j}+W^N_{i,j}\right)$ est valable pour toutes les valeurs de $N$ supérieures $i$.

Cela montre que **pour tout $(i,j)$ tel que $0\leq j \leq i$ et pour tout nombre premier $p>i$, tous les coefficients $\binom{i}{j}$ peuvent s'écrire , comme la somme de deux coefficients dont l'un est un multiple de $p$ et pas l'autre. La parité de la somme $i+j$ permet de déterminer si il s'agit de $S^N_{i,j}$ ou de $W^N_{i,j}$**.

Cette décomposition n'est pas pour autant triviale. Beaucoup des coefficients binomiaux sont supérieurs à $p$, mais aucun n'est divisible par $p$ premier. Néanmoins, ils peuvent tous se décomposer en un terme qui l'est, et un terme qui ne l'est pas, dont on connait les expressions.

Cette décomposition en partie multiple de $p$ et partie non multiple de $p$, n'est pas unique, et ne correspond pas forcément aux termes que l'on obtiendrait en calculant les termes de la décomposition en base $p$. Rien ne garantit par exemple que le terme qui n'est pas congru à $p$ soit positif et inférieur à $p$, car par construction $S^N_{i,j}>0$, mais $W^N_{i,j}$ ne l'est pas forcément.

Pour rappel, on a les symétries suivantes, valables pour $0\leq i \leq N-1$ et $0\leq j \leq N-1$:
- Si $j>i$, 
    - $S^N_{i,j}=S^N_{j,i}$
    - $W^N_{i,j}=W^N_{j,i}$
- Si $i+j>N-1$, 
    - $S^N_{i,j}=S^N_{N-1-j,N-1-i}$
    - $W^N_{i,j}=-W^N_{N-1-j,N-1-i}$

## Concernant les nombres de Fibonacci et de Lucas

Le Triangle de Pascal est directement lié aux nombres de Fibonacci via une somme sur certaines de ses lignes. 

L'utilisation de la décomposition du Damier dans [l'application aux nombres de Fibonacci et de Lucas](./fibonacci_FR.ipynb) a montré que les propriétés des coefficients $S^N_{i,j}$ étaient directement liées à celles des nombres de Lucas.

La décomposition utilisée ici a permis de démontrer une propriété bien connue ($p$ premier $\Rightarrow L_p-1\equiv 0\mod p$), mais également d'étudier cette propriété pour les nombres pseudo-premiers qui correspondent au cas où $N$ n'est pas premier et $\Rightarrow L_N-1\equiv 0\mod N$.

Pour les nombres premiers, l'égalité $L_p-1\equiv 0\mod p$ est une égalité terme à terme lorsqu'on écrit les nombres de Lucas sous la forme d'une somme de terme $S^p_{i,j}$ (ceux tels que $i+j=N-2$). Pour les nombres pseudo premiers, $L_N-1\equiv 0\mod N$ mais on a démontré qu'il existe au moins un terme non congru à 0 modulo N dans cette somme.

## Formules de récurrence

Nous avons démontré plusieurs propriétés de récurrence reliant les $S^N_{i,j}$ et les $W^N_{i,j}$ dont

$$S^N_{i,j}=S^N_{i{-}1,j{-}1}+W^N_{i{-}1,j}$$

et la formule qui en dérive par récurrence pour $3\leq i < N-1$ et $ 1 \leq j \leq i-2$, en posant $q=i-j\ge 2$. 

$$S^N_{i,j}=S^N_{i-1,j-1}+\sum_{k=0}^{j-1}S^N_{i-2-k,j-k}+W^N_{q-1,0}$$

Ces formules sont à $N$ constant, mais $i$ et $j$ variables. 

On peut en dériver d'autres pour lesquels $N$ est également variable (voir [formulaires](./formulaires_FR.ipynb))


Par exemple pour passer de $N$ à $N+1$, on peut définir une règle de construction par récurrence pour tous les $N\ge 1$ et pour tous les i,j telque $0\leq j \leq i \leq N-1$ ne dépendant pas explicitement des coefficients binomiaux, et s'appuyant sur les symétries précédentes:

- si $i=N-1$ et $j=0$ alors $S^N_{N-1,0}=2$ et $W^N_{N-1,0}=0$ (initialisation de la suite)
- sinon
   - si $j>i$, alors (première symétrie)
       - $S^N_{i,j}=S^N_{j,i}$  
       - $W^N_{i,j}=W^N_{j,i}$
    - sinon $j\leq i$ :
        - si $i+j>N-1$ alors (seconde symétrie)
            - $S^N_{i,j}=S^N_{N-1-j,N-1-i}$ 
            - $W^N_{i,j}=-W^N_{N-1-j,N-1-i}$
        - sinon $i+j\leq N-1$:
$$
\begin{eqnarray}
S^{N}_{i,j}&=&{1\over 2}S^{N-1}_{i,j}\left(1+{N-1-j \over N-1-i }\right)+{1\over 2}W^{N-1}_{i,j}\left(1-{N-1-j \over N-1-i }\right)\\
W^{N}_{i,j}&=&{1\over 2}S^{N-1}_{i,j}\left(1-{N-1-j \over N-1-i }\right)+{1\over 2}W^{N-1}_{i,j}\left(1+{N-1-j \over N-1-i }\right)
\end{eqnarray}
$$


La suite par récurrence (à trois indices $i,j,N$) est initialisée pour tout $N\ge 1$ avec $S^N_{N-1,0}=2$ et $W^N_{N-1,0}=0$

Au sens strict, cette formulation n'est pas plus simple que l'expression à partir des coefficients binomiaux, mais elle montre qu'il est possible d'exprimer les termes $S^N_{i,j}$ et $W^N_{i,j}$, à partir de suites à 3 indices définies par récurrence, ne faisant pas apparaître explicitement les coefficients binomiaux.


## Cas $N=p^n$

La [généralisation à $N=p^n$](./p_puissance_n_FR.ipynb) avec $p$ un nombre premier strictement supérieur à 2, a montré qu'il était possible de construire les matrices $S^{p^n}_{i,j}$ et $W^{p^n}_{i,j}$ modulo $p$, par récurrence à partir à partir de $S^{p^{n-1}}_{i',j'}$, $W^{p^{n-1}}_{i',j'}$, et des décompositions en base $p$ de $i=i_{n-1}p^{n-1}+i'$ et $j=j_{n-1}p^{n-1}+j'$ en distinguant les cas suivants :
- $j\leq i$
    - $j'>i \Rightarrow S^{p^n}_{i,j}\equiv0 \mod p$
    - $j'\leq i'$
        - $i_{n-1}+j_{n-1}\equiv 0\mod 2 \Rightarrow S^{p^n}_{i,j}\equiv 2^{-1}S^p_{i_{n-1},j_{n-1}}S^{p^{n-1}}_{i',j'} \mod p$ 
        - $i_{n-1}+j_{n-1}\equiv 1\mod 2 \Rightarrow S^{p^n}_{i,j}\equiv 2^{-1}W^p_{i_{n-1},j_{n-1}}W^{p^{n-1}}_{i',j'} \mod p$ 
- $j>i \Rightarrow S^{p^n}_{i,j}\equiv S^{p^n}_{j,i} \mod p$


## Décomposition alternative $S^N_{i,j},W^N_{i,j}\Rightarrow X^N_{i,j},Y^N_{i,j}$

Le critère de parité pour distinguer les termes congrus à $p$, n'est pas forcément le plus simple à manipuler, même si c'est celui qui est naturel géométriquement. On peut définir une décomposition alternative séparant explicitement les termes congrus de ceux qui ne le sont pas lorsque $N=p$ premier, toujours pour $0\leq j \leq i \leq N-1$.

$$X^N_{i,j}=\binom{i}{j}-(-1)^{i+j}\binom{N-1-j}{N-1-i}$$

$$Y^N_{i,j}=\binom{i}{j}+(-1)^{i+j}\binom{N-1-j}{N-1-i}$$

**Si $N$ est un nombre premier $p$**, alors pour tout $0\leq j\leq i \leq p-1$, $X^N_{i,j}$ est un multiple de $p$, et $Y^N_{i,j}$ ne l'est pas, et on peut exprimer $A_{i,j}$ sous la forme
$$A_{i,j}={1\over 2}\left({X^p_{i,j}\over p}p+Y^p_{i,j}\right)=\binom{i}{j}$$

On peut également en déduire  que pour $0\leq j \leq i \leq p-1$
, $\binom{i}{j}\equiv 2^{-1}Y^p_{i,j} \mod p\not\equiv 0\mod p$

Cette expression est vraie pour tous les nombres premiers $p>i$.

## Généralisation des coefficients binomiaux

Le coefficient binomial $\binom{i}{j}$ est défini pour $i$ et $j$ appartenant $\mathbb{N}$ (avec la convention $\binom{i}{j}=0$ si $j>i$). Ils se généralisent notamment en étendant leur définition au cas $i\in\mathbb{Z}$, via l'expression suivante pour $i<0$. 

On note ce cas $i=-n$, avec $n\in\mathbb{N}^*$ and $k\in\mathbb{N}$:

$$\binom{-n}{k}=(-1)^k\binom{n+k-1}{k}$$

On peut alors exprimer les $X^N_{i,j}$ et $Y^N_{i,j}$, pour tout $0\leq j \leq i \leq N-1$ sous la forme :
$$
\begin{eqnarray}
X^N_{i,j}&=&\binom{i}{i-j}-\binom{i-N}{i-j}\\
Y^N_{i,j}&=&\binom{i}{i-j}+\binom{i-N}{i-j}\\
\end{eqnarray}
$$

qui ne dépend que de la différence $i-j$, et de la décomposition $(i,N-i)$.


## Lien avec les formules de décomposition

La démonstration proposée des propriétés du damier de Pascal, repose sur 
- une formule de récurrence reliant entre elles les cellules sur les diagonales $i-j=Cte$, qui induit une partition entre les cellules selon la parité de $i+j$, puis
- la propriété D qui dit que si $p$ est premier alors $\binom{p}{j}\equiv 0\mod p$ pour tout $1\leq j \leq p-1$.

Dans Rowland (2022), on peut trouver une généralisation du théorème de Lucas modulo $p^2$. L'article se limite à $p$ premier. Le corollaire 7 est un cas particulier de la proposition 6 qui s'appuie sur le développement des produits factoriels des coefficients binomiaux. Le corollaire 7 prouve  que si $p$ premier et  $0\leq s \leq r \leq p-1$, alors 

$$\binom{r}{s}\equiv (-1)^{r-s}\binom{p-1-s}{r-s}\mod p$$. 

Cette expression permet de démontrer la propriété $G$ du damier de Pascal (régularité dans le cas $N$ premier) sans recourir à la formule de récurrence que nous avons utilisée : $$S^N_{i,j}=S^N_{i-1,j-1}+\sum_{k=0}^{j-1}S^N_{i-2-k,j-k}+W^N_{q-1,0}$$.

L'expression découlant du corollaire 7 de Rowland (2022) apparaît également dans Cinkr and Ozturkalan (2023), mais se base sur une démonstration différente de celle de Rowland (2022), et de celle que nous avons utilisée.  La démonstration découle d'une formule de décomposition des coefficients binomiaux :

$$\binom{n}{i}\binom{i}{j}=\binom{n}{j}\binom{n-j}{n-i},$$
pour $n=N-1$, et des propriétés de la ligne $N-1$ du triangle de Pascal pour $N=p$ premier, pour laquelle $\binom{p-1}{j}\equiv (-1)^j\mod p$.


Contrairement à Rowland (2022), qui se limite au cas $N=p$ premier, Cinkr and Ozturkalan (2023), utilise cette propriété pour dériver de nouvelles identités entre coefficients binomiaux.
Certaines découlent de l'utilisation de la formule de Vandermonde et ne sont pas spécifiques aux cas des nombres premiers (Théorème 3.1). D'autres utilisent spécifiquement les propriétés de congurence connues des nombres de Fibonacci $F_p$ et $F_{p+1}$ avec $p$ premier (Théorème 4.3).

Aucune de ces études ne généralise la décomposition des coefficients binomiaux en deux termes $(S,W)$ ou $(X,Y)$ comme nous l'avons proposé, ni ne fait le lien avec les propriétés des nombres des Lucas.


## Structure de corps $\mathbb{Z}/p\mathbb{Z}$

A partir de l'expression générale, on peut écrire plusieurs expressions équivalentes des termes $S^N_{i,j}$ et $W^N_{i,j}$ pour $0\leq j < i \leq N-1$ (Voir [Formulaire](./formulaires_FR.ipynb)).

Par exemple,

$$
S^N_{i,j}=\binom{i}{j}\left(1+\prod_{k=j+1}^{i} {N-k\over k}\right)
$$

Si $p$ est un nombre premier, alors $\mathbb{Z}/p\mathbb{Z}$ est un corps. 
En particulier, pour tout $k\not\equiv 0\mod p$, il existe un inverse $k^{-1}\in\mathbb{Z}/p\mathbb{Z}$
tel que $k^{-1}k\equiv 1\mod p$.

On peut alors développer l'expression précédente.

$$
\begin{eqnarray}
S^p_{i,j}&\equiv&\binom{i}{j}\left(1+\prod_{k=j+1}^{i} k^{-1}(p-k)\right)\mod p\\
&\equiv&\binom{i}{j}\left(1+\prod_{k=j+1}^{i} (k^{-1}p-1)\right)\mod p\\
&\equiv&\binom{i}{j}\left(1+(-1)^{j-i}\right)\mod p\\
\end{eqnarray}
$$

En utilisant $\binom{i}{j}\not\equiv 0\mod p$ pour tout $0\leq j \leq i \leq p-1$, on retrouve bien la propriété des cases blanches et des cases noires du damier lorsque $N=p$ premier.

## Discussion sur les symétries du triangle de Pascal et les valuations p-adiques

L'existence d'une structure fractale triangulaire dans le triangle de Pascal modulo un nombre premier est connue et discutée dans de nombreuses études. Prunescu (2022) étudie entre autres, la symétrie triangulaire du tableau des valuations p-adique $v_p\left(\binom{a}{b}\right)$ d'un triangle de Pascal de dimension $p^m$. La valuation p-adique $v_p(x)$ est la plus grande puissance d'un nombre premier $p$ qui divise un nombre $x$. L'auteur nomme ce triangle, le triangle de Pascal-Kummer, en référence au [théorème de Kummer](https://fr.wikipedia.org/wiki/Th%C3%A9or%C3%A8me_de_Kummer_(coefficients_binomiaux)) qui permet d'exprimer $v_p\left(\binom{a}{b}\right)$ en fonction de la décomposition en base $p$ de $a$, $b$ et $a-b$. Ce théorème se démontre à partir de l'expression des coefficients binomiaux et de la formule de Legendre qui donne la valuation p-adique de $n!$.

Pour $0\leq j \leq i \leq p-1$ avec $p$ premier, aucun des $\binom{i}{j}$ n'est congru à $p$. Cela revient à dire que $v_p\left(\binom{i}{j}\right)=0$. On sait également que $0\leq j \leq i \leq p^m-1$, $v_p\left(\binom{i}{j}\right)\leq m-1$.


Cinkr and Ozturkalan (2023), mais également Rowland (2022) avec des approches différentes, traitent des symétries dans le triangle de Pascal. Ils montrent que la symétrie triangulaire se retrouve dans les propriétés de congruence du triangle. Ces études rentrent dans le cadre plus générale des travaux qui, en étudiant les propriétés de ces coefficients binomiaux modulo $p^m$, visent à généraliser le théorème de Lucas qui relie la décomposition de $i$ et $j$ en base $p$ et la valeur de $\binom{i}{j}\mod p$.

La question générale reste celle des propriétés de la décomposition des coefficients binomiaux $\binom{i}{j}$ en base $p$. Certaines de ces études se focalisent donc sur les propriétés de $\binom{i}{j}\mod p^m$, et d'autres sur la plus grande puissance $p^{v_p}$ telle que $\binom{i}{j}\equiv 0 \mod p^{v_p}$. 
Mais toutes ces études montrent que la symétrie triangulaire du triangle de Pascal joue un rôle dans la compréhension des propriétés des coefficients binomiaux.


# Conclusion 

La construction du damier de Pascal, et l'observation de la conjecture initiale, sont basées sur un repliement du triangle de Pascal selon la symétrie $(i,j)\to(N-1-j,N-1-i)$.

Cela nous a permis de proposer plusieurs décompositions des coefficients binomiaux $\binom{i}{j}$, définies pour toutes les valeurs de $N>i$. 

Lorsque $N$ est un nombre premier $p$, ces décompositions permettent d'obtenir un terme congru à $p$ et un terme non congru à $p$, pour tout $0\leq j \leq i \leq p-1$.

Nous avons proposé deux décompositions. 

## Première décomposition

**La première décomposition** est celle définissant le damier de Pascal et son miroir pour $N>2$, et $0\leq j \leq i \leq N-2$:

$$
\begin{eqnarray}
S^N_{i,j}&=&\binom{i}{j}+\binom{N-1-j}{N-1-i}\\
W^N_{i,j}&=&\binom{i}{j}-\binom{N-1-j}{N-1-i}\\
\end{eqnarray}
$$

Cette décomposition permet la reconstruction

$$
\begin{eqnarray}
\binom{i}{j}&=&{S^N_{i,j}+W^N_{i,j}\over 2}\\
\binom{N-1-j}{N-1-i}&=&{S^N_{i,j}-W^N_{i,j}\over 2}
\end{eqnarray}
$$

Le théorème du damier de Pascal permet de dire que le damier et son miroir modulo $N$ sont réguliers (i.e. sont une alternance parfaite de termes congrus et non congrus, selon la parité de $i+j$), si et seulement si $N>2$ est premier.

Grâce à cette décomposition on montre par exemple que:
- La somme des termes $S^N_{i,j}$ tels que $0\leq j \leq i\leq N-2$ et $i+j=N-2$ est le nombre de Lucas d'indice $N$ :

$$L_N=1+\sum_{\substack{0\leq j \leq i \leq N-1\\ i+j = N-2}}S^N_{i,j}$$

- Pour $p>2$ premier, $L_p-1\equiv 0\mod p$ résulte d'une égalité vraie terme à terme, où tous les termes $S^p_{i,j}$ apparaissant dans la somme ci-dessus sont congrus à 0 modulo $p$. 

- Les nombres pseudo-premiers de Fibonacci, sont les nombres composés tels que $L_N-1\equiv 0\mod N$, mais pour lesquels l'égalité précédente n'est pas vraie terme à terme. Il existe toujours au moins un terme non congru à $0$ dans cette somme.

## Deuxième décomposition

**La seconde décomposition** utilise les coefficients binomiaux généralisés et permet de simplifier les écritures tout en partitionnant les termes congrus à $p$ selon la parité de $i+j$. Pour tous les $0\leq j \leq i \leq N-1$ :

$$
\begin{eqnarray}
X^N_{i,j}&=&\binom{i}{i-j}-\binom{i-N}{i-j}\\
Y^N_{i,j}&=&\binom{i}{i-j}+\binom{i-N}{i-j}
\end{eqnarray}
$$

où $\binom{i-N}{i-j}$ est le coefficient binomial généralisé dont l'indice $i-N$ est négatif, et l'indice $i-j$ est positif ou nul.

- Cette décomposition permet la reconstruction des coefficients binomiaux selon les mêmes expressions que pour $(S,W)$ :

$$
\begin{eqnarray}
\binom{i}{j}&=&{X^N_{i,j}+Y^N_{i,j}\over 2}\\
\binom{i-N}{i-j}&=&{X^N_{i,j}-Y^N_{i,j}\over 2}=(-1)^{i-j}\binom{N-1-j}{N-1-i}
\end{eqnarray}
$$

- Le théorème du damier de Pascal permet de dire $X^N_{i,j}\equiv 0\mod N$ et $Y^N_{i,j}\not\equiv 0\mod N$ pour tout $0\leq j \leq i \leq N-1$ si et seulement si $N>2$ et $N$ est premier. C'est à dire que pour $N=p>2$ premier, on peut en déduire que pour tout $0\leq j \leq i \leq p-1$ :

$$
\begin{eqnarray}
X^p_{i,j}&\equiv&0\mod p\\
\binom{i}{j}&\equiv& 2^{-1}Y^p_{i,j}\mod p\\
&\not\equiv& 0\mod p \Rightarrow  Y^p_{i,j}\not\equiv 0\mod p\\
\end{eqnarray}
$$

- En utilisant cette décomposition et l'identité de Vandermonde, il est possible de dériver plusieurs propriétés montrant par exemple que $\binom{p+i}{p+j}\equiv\binom{i}{j}\mod p$, ou encore que pour tout $0\leq j < i \leq N-1$ et pour tout $u$ et $v$ tels que $u+v=N$, 

$$
\begin{eqnarray}
X^{N=u+v}_{i,j}&=&\sum_{m=1}^{i-j}\binom{i-u}{i-j-m}\left[\binom{u}{m}-\binom{-v}{m}\right]\\
&=&\sum_{m=1}^{i-j}\binom{i-v}{i-j-m}\left[\binom{v}{m}-\binom{-u}{m}\right]\\
\end{eqnarray}
$$

Le terme de gauche ne dépend que de $N$. Cette identité est donc vraie pour tous les couples $(u,v)$ et $(u',v')$ et pour tout $0\leq j < j \leq N-1$, dès lors que $u+v=u'+v'=N$. Le nombre de termes dans la somme ne dépend que de $i-j$ (NB: si $i=j=k$, $X^N_{k,k}=0$). De plus, si $N$ est un nombre premier $p$ supérieur à 2, alors cette somme est congrue à 0 modulo $p$.

## Triangle et damiers 
On résume ici brièvement les définitions 

### Triangle de Pascal

Pour $0\leq j \leq i\leq N-1$, les coefficients $A_{i,j}$ et $B^N_{i,j}$ sont définis à partir des coefficients binomiaux des $N$ premières lignes du triangle de Pascal :

$$\begin{eqnarray}
A_{i,0}&=&\binom{i}{0}=1\\
A_{i,i}&=&1\\
A_{i,j}&=&\binom{i}{j}=\binom{i}{i-j}=A_{i,i-j}\\
&=&A_{i-1,j-1}+A_{i-1,j}\\
\end{eqnarray}$$

Pour $0\leq i < j \leq N-1$, $A_{i,j}=A_{j,i}$.

Les coefficients $B^N_{i,j}$ de $B$ sont obtenus par rotation et symétrie de $A$, et dépendent de $N$.

Pour $0\leq j \leq i \leq N-1$ :

$$\begin{eqnarray}
B^N_{i,0}&=&\binom{N-1}{i}\\
B^N_{i,i}&=&1\\
B^N_{i,j}&=&\binom{N-1-j}{N-1-i}=\binom{N-1-j}{i-j}=B^N_{N-1-i+j,j}\\
&=&B^N_{i-1,j-1}-B^N_{i-1,j}
\end{eqnarray}
$$

Pour $0\leq i < j \leq N-1$, $B^N_{i,j}=B^N_{j,i}$

Cette formulation est équivalente à $B^N_{i,j}=A_{N-1-j,N-1-i}$, pour $0\leq i,j \leq N-1$.

### Damier SW

Les $S^N_{i,j}$ et $W^N_{i,j}$ sont des coefficients définis, pour $0\leq j \leq i \leq N-1$ par les relations :

$$\begin{eqnarray}
S^N_{i,0}&=&1+\binom{N-1}{i}\\
S^N_{k,k}&=&2\\
S^N_{i,j}&=&A_{i,j}+B^N_{i,j}=\binom{i}{j}+\binom{N-1-j}{N-1-i}\\
&=&S^N_{i{-}1,j{-}1}+W^N_{i{-}1,j}\\
\\
W^N_{i,0}&=&1-\binom{N-1}{i}\\
W^N_{i,i}&=&0\\
W^N_{i,j}&=&A_{i,j}-B^N_{i,j}=\binom{i}{j}-\binom{N-1-j}{N-1-i}\\
&=&W^N_{i{-}1,j{-}1}+S^N_{i{-}1,j}\\
\end{eqnarray}$$

Toujours avec les symétries $S^N_{i,j}=S^N_{j,i}$ et $W^N_{i,j}=W^N_{j,i}$, pour $0\leq i < j \leq N-1$.

### Damier XY

Les $X^N_{i,j}$ et $Y^N_{i,j}$ sont des coefficients définis, pour $0\leq j \leq i \leq N-1$ par les relations :
$$\begin{eqnarray}
X^N_{i,j}&=&\binom{i}{i-j}-\binom{i-N}{i-j}\\
&=&X^N_{i-1,j-1}+X^N_{i-1,j}\\
\\
Y^N_{i,j}&=&\binom{i}{i-j}+\binom{i-N}{i-j}\\
\\&=&Y^N_{i-1,j-1}+Y^N_{i-1,j}\\
\end{eqnarray}$$

Les règles de décomposition sont identiques à celle des coefficients binomiaux du triangle de Pascal, mais les conditions initiales sont différentes 
$$\begin{eqnarray}
X^N_{i,0}&=&1-\binom{i-N}{i}\\
X^N_{i,i}&=&0\\
Y^N_{i,0}&=&1+\binom{i-N}{i}\\
Y^N_{i,i}&=&2
\end{eqnarray}$$

Pour $0\leq j < i \leq N-1$, on utilise  la symétrie $X^N_{i,j}=X^N_{j,i}$ et $Y^N_{i,j}=Y^N_{j,i}$ 

Les $X^N_{i,j}$ et $Y^N_{i,j}$ ont été étudiés avec la condition $0\leq j \leq i \leq N-1$, mais avec cette définition et l'utilisation des coefficients binomiaux généralisés, les $i$ et $j$ ne sont plus obligés d'être inférieurs à N.
