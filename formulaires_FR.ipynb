## Formules de récurrence $(S,W)$

Nous avons démontré plusieurs propriétés de récurrence reliant les $S^N_{i,j}$ et les $W^N_{i,j}$ dont

$$S^N_{i,j}=S^N_{i{-}1,j{-}1}+W^N_{i{-}1,j}$$

qui est valable pour $0\leq j \leq i \leq N-1$. 

La formule qui en dérive par récurrence pour $3\leq i \leq N-1$ et $ 1 \leq j \leq i-2$, en posant $q=i-j\ge 2$. 


$$S^N_{i,j}=S^N_{i-1,j-1}+\sum_{k=0}^{j-1}S^N_{i-2-k,j-k}+W^N_{q-1,0}$$

Ces formules sont établies à $N$ constant. On peut en dériver d'autres à $N$ variable. On pose $N=N'+\Delta$.

Exprimer $S^N_{i,j}$ en fonction de $S^{N'}_{i,j}$ pour $0\leq j < i \leq N'-1$ ne pose pas vraiment de difficulté
(On exclut $i=j$ car $S^N_{i,j}=S^{N'}_{i,i}=2$)


$$
\begin{eqnarray}
S^N_{i,j}&=&\binom{i}{j}+\binom{N'-1+\Delta-j}{N'-1+\Delta-i}\\
         &=&\binom{i}{j}+\binom{N'-1-j}{N'-1-i}\prod_{k=0}^{\Delta-1}{N'-1+\Delta-j-k \over N'-1+\Delta-i-k }\\
         &=&\binom{i}{j}+\binom{N'-1-j}{N'-1-i}\prod_{k=1}^{\Delta}{N'-1-j+k \over N'-1-i+k }\\
         &=&\binom{i}{j}+\binom{N'-1-j}{N'-1-i}\prod_{k=N'}^{N-1}{k-j \over k-i }
\end{eqnarray}
$$

$$
\begin{eqnarray}
S^N_{i,j}&=&\binom{i}{j}+\binom{N'-1+\Delta-j}{N'-1+\Delta-i}\\
         &=&\binom{i}{j}+\binom{N'-1+\Delta-j}{N'-1+\Delta-i}{(N'-i)!\over(N'-j)!}{(N-j)!\over(N-i)!}\\
         &=&\binom{i}{j}+\binom{N'-1-j}{N'-1-i}\prod_{k=j+1}^{i}{N-k \over N'-k }
\end{eqnarray}
$$

$$
\begin{eqnarray}
S^N_{i,j}&=&{1\over 2}(S^{N'}_{i,j}+W^{N'}_{i,j})+{1\over 2}(S^{N'}_{i,j}-W^{N'}_{i,j})\prod_{k=j+1}^{i}{N-k \over N'-k }\\
         &=&{1\over 2}S^{N'}_{i,j}\left(1+\prod_{k=j+1}^{i}{N-k \over N'-k }\right)+{1\over 2}W^{N'}_{i,j}\left(1-\prod_{k=j+1}^{i}{N-k \over N'-k }\right)\\
         &=&{1\over 2}S^{N'}_{i,j}\left(1+\prod_{k=N'}^{N-1}{k-j \over k-i }\right)+{1\over 2}W^{N'}_{i,j}\left(1-\prod_{k=N'}^{N-1}{k-j \over k-i }\right)
\end{eqnarray}
$$

$$
\begin{eqnarray}
W^N_{i,j}&=&{1\over 2}(S^{N'}_{i,j}+W^{N'}_{i,j})-{1\over 2}(S^{N'}_{i,j}-W^{N'}_{i,j})\prod_{k=j+1}^{i}{N-k \over N'-k }\\
         &=&{1\over 2}S^{N'}_{i,j}\left(1-\prod_{k=j+1}^{i}{N-k \over N'-k }\right)+{1\over 2}W^{N'}_{i,j}\left(1+\prod_{k=j+1}^{i}{N-k \over N'-k }\right)\\
         &=&{1\over 2}S^{N'}_{i,j}\left(1-\prod_{k=N'}^{N-1}{k-j \over k-i }\right)+{1\over 2}W^{N'}_{i,j}\left(1+\prod_{k=N'}^{N-1}{k-j \over k-i }\right)
\end{eqnarray}
$$



Exemple pour passer de $N$ à $N+1$ (attention l'expression précédente permet de passer de $N-1$ à $N$).

On connaît les $S^{N}_{i,j}$ et $W^{N}_{i,j}$ pour tous les $0\leq j \leq i \leq N-1$. 

On cherche les $S^{N+1}_{i,j}$ et $W^{N+1}_{i,j}$ pour tous les $0\leq j \leq i \leq N$.


La forme précédente nous les donne tous sauf pour $i=N$
$$S^{N+1}_{i,j}={1\over 2}S^{N}_{i,j}\left(1+{N-j \over N-i }\right)+{1\over 2}W^{N}_{i,j}\left(1-{N-j \over N-i }\right)$$

$$W^{N+1}_{i,j}={1\over 2}S^{N}_{i,j}\left(1-{N-j \over N-i }\right)+{1\over 2}W^{N}_{i,j}\left(1+{N-j \over N-i }\right)$$


Pour $i=N$, on l'obtient simplement $S^{N+1}_{N,j}=1+\binom{N}{j}$. On peut se passer des coefficients binomiaux, en utilisant la formule de récurrence à $N$ fixe , en démarrant avec $S^{N+1}_{N,0}=2$ et $W^{N+1}_{N,0}=0$, tous les autres termes étant connus, ou encore simplement utiliser la symétrie pour $i+j<N-1$ qui donne $S^N_{i,j}=S^N_{N-1-j,N-1-i}$, qui fait qu'il ne nous manque que $S^{N+1}_{N,0}=2$, $W^{N+1}_{N,0}=0$, 

Et on a donc une règle de construction par récurrence pour tous les $N>1$ et pour tous les i,j tel que $0\leq j \leq i < N-1$). 

On peut donc généraliser l'expression en prenant en compte les différentes symétries.


- si $i=N-1$ et $j=0$ alors (initialisation)
    - $S^N_{N-1,0}=2$
    - $W^N_{N-1,0}=0$ 
- sinon
   - si $j>i$, alors  (symétrie #1)
       - $S^N_{i,j}=S^N_{j,i}$  
       - $W^N_{i,j}=W^N_{j,i}$
    - sinon $j\leq i$ :
        - si $i+j>N-1$ alors (symétrie #2)
            - $S^N_{i,j}=S^N_{N-1-j,N-1-i}$ 
            - $W^N_{i,j}=-W^N_{N-1-j,N-1-i}$
        - sinon $i+j\leq N-1$:
$$
\begin{eqnarray}
S^{N}_{i,j}&=&{1\over 2}S^{N-1}_{i,j}\left(1+{N-1-j \over N-1-i }\right)+{1\over 2}W^{N-1}_{i,j}\left(1-{N-1-j \over N-1-i }\right)\\
W^{N}_{i,j}&=&{1\over 2}S^{N-1}_{i,j}\left(1-{N-1-j \over N-1-i }\right)+{1\over 2}W^{N-1}_{i,j}\left(1+{N-1-j \over N-1-i }\right)
\end{eqnarray}
$$


La suite par récurrence (à trois indices $i,j,N$) est initialisée pour tout $N\ge 1$ avec $S^N_{N-1,0}=2$ et $W^N_{N-1,0}=0$

Au sens strict, cette formulation n'est pas plus simple que l'expression à partir des coefficients binomiaux, mais elle montre qu'il est possible d'exprimer la décomposition en termes $S^N_{i,j}$ et $W^N_{i,j}$ comme des suites à 3 indices définies par récurrence.


On peut utiliser une formulation légèrement différente pour s'assurer que les termes sont toujours des entiers.

$$
\begin{eqnarray}
S^{N}_{i,j}&=&{\left(2(N-1)-(i+j)\right)S^{N-1}_{i,j}+(j-i)W^{N-1}_{i,j}\over 2(N-1-i)}\\
W^{N}_{i,j}&=&{(j-i)S^{N-1}_{i,j}+\left(2(N-1)-(i+j)\right)W^{N-1}_{i,j}\over 2(N-1-i)}
\end{eqnarray}
$$

## Décomposition alternative $(X,Y)$

Le cas échéant, cela permet de formuler une expression alternative séparant explicitement les termes congrus à $p$ des termes qui ne le sont pas. 

$$X^N_{i,j}=\binom{i}{j}-(-1)^{i+j}\binom{N-1-j}{N-1-i}={1-(-1)^{i+j}\over 2}S^N_{i,j}+{1+(-1)^{i+j}\over 2}W^N_{i,j}$$

$$Y^N_{i,j}=\binom{i}{j}+(-1)^{i+j}\binom{N-1-j}{N-1-i}={1+(-1)^{i+j}\over 2}S^N_{i,j}+{1-(-1)^{i+j}\over 2}W^N_{i,j}$$


Par symétrie, l'expression de $X^N_{i,j}$ et $Y^N_{i,j}$ en fonction de $S^N_{i,j}$ et $W^N_{i,j}$ est vraie pour $0\leq i,j \leq N-1$

Si $N$ est un nombre premier $p$, alors $X^N_{i,j}$ est un multiple de $p$, et $Y^N_{i,j}$ ne l'est pas. 

C'est à dire que pour $0\leq i,j \leq p-1$, on peut toujours mettre $A_{i,j}$ sous la forme

$$A_{i,j}={1\over 2}\left({X^p_{i,j}\over p}p+Y^p_{i,j}\right)$$

**NB** : Il s'agit bien ici de la matrice $A_{i,j}$, et pas du coefficient binomial. Pour $0\leq j \leq i \leq N-1$, $A_{i,j}=\binom{i}{j}=A_{j,i}$, alors que $\binom{j}{i}=0$ pour tout $j>i$.


De façon similaire à ce qu'on a fait pour $S$ et $W$, on peut développer l'expression par récurrence de $X^N_{i,j}$ et $Y^N_{i,j}$ en fonction de $X^{N-1}_{i,j}$ et $Y^{N-1}_{i,j}$.  

Les symétries ne sont pas les mêmes que pour $S$ et $W$.

Pour $N>1$ et tous les i,j tels que $0\leq j,i \leq N-1$.


- si $i=N-1$ et $j=0$ alors (initialisation)
    - $X^N_{N-1,0}=1-(-1)^{N-1}$ (=2 si $N$ est pair, =0 sinon)
    - $Y^N_{N-1,0}=1+(-1)^{N-1}$ (=0 si $N$ est pair, =2 sinon)
- sinon
   - si $j>i$, alors  (symétrie #1)
       - $X^N_{i,j}=X^N_{j,i}$  
       - $Y^N_{i,j}=Y^N_{j,i}$
    - sinon $j\leq i$ :
        - si $i+j>N-1$ alors (symétrie #2)
            - $X^N_{i,j}=|X^N_{N-1-j,N-1-i}|$ 
            - $Y^N_{i,j}=|Y^N_{N-1-j,N-1-i}|$
        - sinon $i+j\leq N-1$:
$$
\begin{eqnarray}
X^{N}_{i,j}&=&{1\over 2}X^{N-1}_{i,j}\left(1+{N-1-j \over N-1-i }\right)+{1\over 2}Y^{N-1}_{i,j}\left(1-{N-1-j \over N-1-i }\right)\\
Y^{N}_{i,j}&=&{1\over 2}X^{N-1}_{i,j}\left(1-{N-1-j \over N-1-i }\right)+{1\over 2}Y^{N-1}_{i,j}\left(1+{N-1-j \over N-1-i }\right)
\end{eqnarray}
$$

## Généralisation des coefficients binomiaux

Le coefficient binomial $\binom{i}{j}$ est défini pour $i$ et $j$ appartenant $\mathbb{N}$. Ils se généralisent notamment en étendant leur définition au cas $i\in\mathbb{Z}$, via l'expression suivant pour $i<0$. On note ce cas $i=-n$, avec $n\in\mathbb{N}^*$:

$$\binom{-n}{j}=(-1)^j\binom{n+j-1}{j}$$

Cette expression s'obtient directement de l'expression généralisée (https://en.wikipedia.org/wiki/Binomial_coefficient)

*"The multiplicative formula allows the definition of binomial coefficients to be extended by replacing n by an arbitrary number α (negative, real, complex) or even an element of any commutative ring in which all positive integers are invertible:*
$$\binom {\alpha }{k}={\frac {\alpha ^{\underline {k}}}{k!}}={\frac {\alpha (\alpha -1)(\alpha -2)\cdots (\alpha -k+1)}{k(k-1)(k-2)\cdots 1}}$$
*for $k\in\mathbb {N}$ and arbitrary $\alpha$*".

Chaque terme du numérateur est négatif. En factorisant $-1$ pour les $j$ termes du numérateur, on fait apparaître le coefficient binomial $\binom{n+j-1}{j}$.

On peut alors faire la transformation suivante des expressions de $X^N_{i,j}$ et $Y^N_{i,j}$, en posant $i-j=k$:

$$
\begin{eqnarray}
X^N_{i,j}&=&\binom{i}{j}-(-1)^{i+j}\binom{N-1-j}{N-1-i}\\
&=&\binom{i}{i-j}-(-1)^{i-j}\binom{N-1-j}{i-j}\\
&=&\binom{i}{k}-(-1)^k\binom{N-1+k-i}{k}\\
&=&\binom{i}{k}-(-1)^k\binom{N-i+k-1}{k}\\
&=&\binom{i}{k}-\binom{-(N-i)}{k}\\
&=&\binom{i}{k}-\binom{i-N}{k}
\end{eqnarray}
$$



---
**On aboutit alors aux expressions suivantes définissant les décompositions $(S,W)$ et $(X,Y)$ :**

- Pour $0\leq j \leq i \leq N-1$
$$
\begin{eqnarray}
X^N_{i,j}&=&\binom{i}{i-j}-\binom{i-N}{i-j}\\
Y^N_{i,j}&=&\binom{i}{i-j}+\binom{i-N}{i-j}\\
\end{eqnarray}
$$

qui ne dépendent que de $i$, $N-i$, et de $i-j$.

- Les cas $0\leq i < j \leq N-1$ sont donnés par la symétrie $S^N_{i,j}=S^N_{j,i}$.

## En lien avec la formule de Vandermonde pour $N=u+v$

**En posant $N=u+v$** et $v=u-k$, on peut alors écrire ces termes sous la forme $X^{N=u+v}_{i=u,j=u-k}$ et $Y^{N=u+v}_{i=u,j=u-k}$, avec $0\leq k\leq u$ et $0\leq u \leq N-1$:

$$
\begin{eqnarray}
X^{u+v}_{u,u-k}&=&\binom{u}{k}-\binom{-v}{k}\\
Y^{u+v}_{u,u-k}&=&\binom{u}{k}+\binom{-v}{k}\\
\end{eqnarray}
$$

---

En utilisant la formule de Vandermonde, pour $0\leq j \leq i \leq N-1$

$$
\begin{eqnarray}
X^{N}_{i,j}&=&\binom{i}{i-j}-\binom{i-N}{i-j}\\
&=&\binom{i-N+N}{i-j}-\binom{i-N}{i-j}\\
&=&\sum_{m_1+m_2=i-j}\binom{i-N}{m_1}\binom{N}{m_2}-\binom{i-N}{i-j}\\
&=&\sum_{m=0}^{i-j}\binom{i-N}{m}\binom{N}{i-j-m}-\binom{i-N}{i-j}\\
&=&\sum_{m=0}^{i-j-1}\binom{i-N}{m}\binom{N}{i-j-m}\\
\end{eqnarray}
$$

---

pour $0\leq j \leq i \leq N-1$

$$
\begin{eqnarray}
X^{u+v}_{i,j}&=&\binom{i}{i-j}-\binom{i-u-v}{i-j}\\
&=&\binom{i-u+u}{i-j}-\binom{i-u-v}{i-j}\\
&=&\sum_{m_1+m_2=i-j}\binom{i-u}{m_1}\binom{u}{m_2}-\sum_{m_1+m_2=i-j}\binom{i-u}{m_1}\binom{-v}{m_2}\\
&=&\sum_{m=0}^{i-j}\binom{i-u}{m}\left[\binom{u}{i-j-m}-\binom{-v}{i-j-m}\right]\\
&=&\sum_{m=0}^{i-j}\binom{i-u}{i-j-m}\left[\binom{u}{m}-\binom{-v}{m}\right]\\
&=&\sum_{m=1}^{i-j}\binom{i-u}{i-j-m}\left[\binom{u}{m}-\binom{-v}{m}\right]\\
&=&\sum_{m=1}^{i-j}\binom{i-v}{i-j-m}\left[\binom{v}{m}-\binom{-u}{m}\right]\\
\end{eqnarray}
$$

---
pour $0\leq j \leq i \leq N-1$

$$
\begin{eqnarray}
X^{N}_{i,j}&=&\binom{i}{i-j}-\binom{i-N}{i-j}\\
&=&\sum_{m=0}^{i-j-1}\binom{i-N}{m}\binom{N}{i-j-m}\\
&=&\sum_{m=0}^{i-j-1}\left[\binom{i}{m}-X^N_{i,i-m}\right]\binom{N}{i-j-m}\\
&=&\sum_{m=0}^{i-j-1}\binom{i}{m}\binom{N}{i-j-m}-\sum_{m=0}^{i-j-1}X^N_{i,i-m}\binom{N}{i-j-m}\\
&=&\sum_{m=0}^{i-j-1}\binom{i}{m}\binom{N}{i-j-m}-\sum_{m=1}^{i-j}X^N_{i,j+m}\binom{N}{m}\\
&=&\left[-\binom{i}{i-j}\binom{N}{0}+\sum_{m=0}^{i-j}\binom{i}{m}\binom{N}{i-j-m}\right]-\sum_{m=1}^{i-j}X^N_{i,j+m}\binom{N}{m}\\
&=&\left[-\binom{i}{i-j}+\sum_{m=0}^{i-j}\binom{i}{m}\binom{N}{i-j-m}\right]-\sum_{m=1}^{i-j}X^N_{i,j+m}\binom{N}{m}\\
\\
X^{N}_{i,j}&=&\binom{N+i}{N+j}-\binom{i}{j}-\sum_{m=1}^{i-j}X^N_{i,j+m}\binom{N}{m}\\
&=&\binom{N+i}{i-j}-\binom{i}{i-j}-\sum_{m=0}^{i-j-1}X^N_{i,i-m}\binom{N}{i-j-m}\\
\end{eqnarray}
$$

Pour $N=p$ premier, on trouve directement la propriété $\binom{p+i}{i-j}=\binom{p+i}{p+j}\equiv \binom{i}{i-j}\mod p$, que l'on peut associer au triangle modulo $p$ décaler de $p$ lignes et $p$ colonne. Le changement $i\to i-N+N$ peut être généralisé pour dériver d'autres propriétés.

## Lien avec les formules de décomposition (1/2)

Pour tout $i$, $j$, $N$, on a la relation suivante

$$\binom{N}{i}\binom{i}{j}=\binom{N}{j}\binom{N-j}{N-i}$$

$$
\begin{eqnarray}
\binom{N}{i}\binom{i}{j}&=&\frac{N!}{i!(N-i)!}\frac{i!}{j!(i-j)!}\\
&=&\frac{N!}{(N-i)!}\frac{1}{j!(i-j)!}\\
&=&\frac{N!}{j!}\frac{1}{(N-i)!(i-j)!}\\
&=&\frac{N!}{j!(N-j)!}\frac{(N-j)!}{(N-i)!(i-j)!}\\
&=&\binom{N}{j}\binom{N-j}{N-i}
\end{eqnarray}
$$

La relation de congruence correspondante pour pour $N=p$ premier est triviale (pour tous les $1 \leq j\leq N-2$, $\binom{p}{j}\equiv 0\mod p$).

Mais si on écrit la même expression pour la ligne $N-1$, c'est plus intéressant du point de vue du damier et de son miroir, puisqu'on retrouve l'expression des $A_{i,j}$ et $B^N_{i,j}$.

$$\binom{N-1}{i}\binom{i}{j}-\binom{N-1}{j}\binom{N-1-j}{N-1-i}=0$$

et en conséquence

$$
\begin{eqnarray}
(-1)^i\binom{i}{j} -(-1)^j\binom{N-1-j}{N-1-i}\equiv 0\mod p\\
%(-1)^i\binom{i}{j}-(-1)^j\binom{N-1-j}{N-1-i}\equiv 0 \mod p\\
\binom{i}{j}-(-1)^{j-i})\binom{N-1-j}{N-1-i}\equiv 0 \mod p
\end{eqnarray}
$$


Cette propriété se trouve dans la littérature récente (Rowland, 2022; Cinkir and Ozturkalan, 2023; identifiées après l'observation de la conjecture et la première démonstration du théorème du damier), sous la forme

$$\binom{i}{j}\equiv (-1)^{i+j}\binom{p-1-j}{i-j}\mod p$$

C'est à dire que :
- si $i+j\equiv 0 \mod 2$ , on obtient $W^p_{i,j}=\binom{i}{j}-\binom{N-1-j}{N-1-i}\equiv 0\mod p$
- si $i+j\equiv 1 \mod 2$ , on obtient $S^p_{i,j}=\binom{i}{j}+\binom{N-1-j}{N-1-i}\equiv 0\mod p$

Ce qui correspond à une démonstration de le propriété des cases noires du damier et du miroir pour $p$ premier. 

On peut également l'utiliser pour démontrer que pour $N=p$ premier, les autres cases du damier et du miroir sont blanches puisque 

- si $i+j\equiv 0 \mod 2$ , on obtient $S^p_{i,j}\equiv S^p_{i,j}+W^p_{i,j}\equiv 2A_{i,j}\not\equiv 0\mod p$
- si $i+j\equiv 1 \mod 2$ , on obtient $W^p_{i,j}\equiv S^p_{i,j}+W^p_{i,j}\equiv 2A_{i,j}\not\equiv 0\mod p$

Cette démonstration est différente de la première démonstration faite dans le cadre de ce travail, qui repose elle sur la formule de récurrence, rappelée précédemment, reliant les $S^N_{i,j}$ entre eux:

$$S^N_{i,j}=S^N_{i-1,j-1}+\sum_{k=0}^{j-1}S^N_{i-2-k,j-k}+W^N_{q-1,0}$$

## Lien avec les formules de décomposition (2/2)

La formule de décomposition précédente nous permet d'écrire:
$$\binom{N-1}{i}\binom{i}{j}-\binom{N-1}{j}\binom{N-1-j}{N-1-i}=0$$

C'est à dire 
$$\binom{i}{j}={\binom{N-1}{j}\over \binom{N-1}{i}}\binom{N-1-j}{N-1-i}$$

Cette égalité est toujours vraie. 

Les termes de gauche et de droite sont bien des entiers positifs. 

Par contre, pour $0\leq j \leq i \leq N-1$, le terme  ${\binom{N-1}{j}\over \binom{N-1}{i}}={i!(N-1-i)!\over j!(N-1-j)!}$ correspond à une fraction rationnelle qui est 
- égale à 1 si $i-j=0$ ou $i+j=N-1$
- supérieure à 1 si $i+j>N-1$
- inférieure à 1 si $i+j<N-1$

Pour tous les $0\leq j < i \leq N-1$, $i!> j!$ et $(N-1-j)!>(N-1-i)$.

On peut donc simplifier la fraction:
$$
\begin{eqnarray}
{\binom{N-1}{j}\over \binom{N-1}{i}}&=&{i!(N-1-i)!\over j!(N-1-j)!}\\
&=&{i\cdots(j+1)\over (N-1-j)\cdots(N-1-i+1)}\\
&=&{i\cdots(j+1)\over (N-i)\cdots(N-(j+1))}\\
&=&\prod_{k=j+1}^{i} {k\over N-k}
\end{eqnarray}
$$

On obtient donc plusieurs expressions équivalentes des termes $S^N_{i,j}$ et $W^N_{i,j}$ pour $0\leq j < i \leq N-1$ 

$$
\begin{eqnarray}
S^N_{i,j}&=&\binom{i}{j}+\binom{N-1-j}{N-1-i}\\
&=&\binom{i}{i-j}+\binom{N-1-j}{i-j}\\
&=&\binom{i}{j}\left(1+\prod_{k=j+1}^{i} {N-k\over k}\right)\\
&=&\binom{N-1-j}{N-1-i}\left(1+\prod_{k=j+1}^{i}{k\over N-k}\right)\\
&&\\
W^N_{i,j}&=&\binom{i}{j}-\binom{N-1-j}{N-1-i}\\
&=&\binom{i}{i-j}-\binom{N-1-j}{i-j}\\
&=&\binom{i}{j}\left(1-\prod_{k=j+1}^{i} {N-k\over k}\right)\\
&=&\binom{N-1-j}{N-1-i}\left(-1+\prod_{k=j+1}^{i} {k\over N-k}\right)
\end{eqnarray}
$$

Dans le cas où $N=p$ est un nombre premier, tous les termes $k$ ou $N-k$ sont tels que $gcd(k,p)=gcd(N-k,p)=1$, et admet donc un inverse. 

On a alors $kk^{-1}\equiv 1\mod p$ et on obtient 

$$\prod_{k=j+1}^{i} (N-k)k^{-1}\equiv\prod_{k=j+1}^{i} k(N-k)^{-1}\equiv (-1)^{i-j}\mod p$$

qui permet bien de retrouver la propriété précédente de regularité du damier et de son miroir pour $N=p$. 

La démonstration de la récipoque reste quand à elle basée sur l'existence de contre-exemple démontrant que le damier et son miroir ne sont pas réguliers si $N$ n'est pas premier.
